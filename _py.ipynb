```python
##Feature-Based Analysis1.Which patient has the highest bilirubin?2.Which patient has the highest albumin?3.Which patient has the lowest platele?
```

```python
print(df.loc[df['Bilirubin'].idxmax(), ['ID', 'Bilirubin', 'Status', 'Stage', 'N_Days']])

print(df.loc[df['Albumin'].idxmax(), ['ID', 'Albumin', 'Status', 'Stage', 'N_Days']])

print(df.loc[df['Platelets'].idxmin(), ['ID', 'Platelets', 'Status', 'Stage', 'N_Days']])
```
